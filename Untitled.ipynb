```python
def table3(n):
    if n > 10:         
        return 
    print(n*1)
    table3(n+1)  
```

```python
c = 1
def table4(n):
    global c
    if c>10:
        return
    print(n*c)
    c=c+1
    table4(n)
```

```python
table4(5)
```

```python
c = 1
def table6(n):
    global c
    if c == 11:
        return
    print(n * c)
    c += 1
    table6(n)
```
