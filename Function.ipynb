```python
# 1. Function without parameter and no return value.
def first():
    print("Welcome")
```

```python
# Function with parameter and return value.
def withdraw(password,amt):
    if password == 123:
        return amt
    else:
        return "wrong password!"
```

# Positional Arguments

ARBITRARY ELEMENT/ARGUMENT-

    1.VARIABLLE LENGTH POSITIONAL ARGUMENTS
    2.VARIABLE LENGTH KEYWORD ARGUMENTS

```python
def show(name,age):
    print("your name : ",name , "your age : ", age)
    
```

```python
# show("abhi",23)
# show(23,"Abhi")

show(age = 20, name="abhi")
```

```python
def show(*info): # Variable length positional arguments (it always return data in tuple format)
    print(info)

show(["abhi","pari","palak"],[22,20,21],["cs","it","cs"], {"code" : 34})
```

```python
#VARIABLE LENGTH KEYWORD ARGUMENTS

def clg(**kw): # it declared  by **arg and returns dictionary.
    print(kw)

clg(name ="anoop", age = 20)
```

```python
# Using both
def university(*clg,**info):
    print(clg,info)
university("rohan",23,"Bhopal", rollno = 101, branch = "cs") 
```

```python
# Lambda function or anonymous function: Nameless 
L = [2,3,4,5,6]
print(L)
t= tuple(map(lambda x:x*x,L))
print(t)
```

```python
cybrom = lambda x:x.upper()
print(type(cybrom))
```

```python
cybrom("abhi")
```

```python
# even_odd (9,8,10,34,5)
```

```python
num = 2,7,4,13,14,17
print(num)
even_odd = list(map(lambda num : "Even"if num%2==0 else "Odd",num))
print(even_odd)
even = list(filter(lambda num : num % 2 ==0 ,  num))
print(even)
```

```python
num = 2,7,4,13,14,17
is_even = lambda x:x%2==0
for n in num:
    if is_even(n):
        print(n, end=" ")
```

```python
num = 2,7,4,13,14,17
is_odd = lambda x:x%2!=0
for n in num:
    if is_odd(n):
        print(n,end=" ")
```

```python
# Reduce function 
from functools import reduce

numbers = [1,2,3,4]
product = reduce(lambda x,y: x * y, numbers)
print(product)
```
