### Four types of functions
###### *Function without parameter and no return value.*
###### *function without parameter and return value.*
###### *Function with parameter and no return value.*
###### *Function with parameter and return value.*

```python
# 1. Function without parameter and no return value.
def first():
    print("Welcome")
```

```python
a = first()
print(a)
```

```python
# 2. function without parameter and return value. 
def prasaad():
    return "Apple"
```

```python
guruji = prasaad()
print(guruji + " Laddu")
```

```python
# 3. Function with parameter and no return value.
# def result(a = 6):  default argument Use case :- Zero balance account
def result(a): # Use case - View Result                                        
    print(a)
```

```python
# Function with parameter and no return value.
def show(a,b):
    print(a+b)
```

```python
# Function with parameter and return value.
def withdraw(password,amt):
    if password == 123:
        return amt
    else:
        return "wrong password!"
```

```python
withdraw(123, 6000)
```
